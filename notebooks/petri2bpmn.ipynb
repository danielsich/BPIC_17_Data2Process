# Load and transform the models to BPMN
- for easier understanding and decision minining

- import relevant libraries and petri nets

```python
import pm4py
import os
```

- load in the intresting petri nets 

```python
heuristic_30_net, heuristic_30_initial_marking, heuristic_30_final_marking = pm4py.read_pnml("../petrinets/heuristic_variants/petri_model_30.pnml")
heuristic_30_bpmn_graph = pm4py.convert_to_bpmn(heuristic_30_net, heuristic_30_initial_marking, heuristic_30_final_marking)
pm4py.view_bpmn(heuristic_30_bpmn_graph)
```
