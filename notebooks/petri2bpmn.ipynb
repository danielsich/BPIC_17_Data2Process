# Load and transform the models to BPMN
- for easier understanding and decision minining

- import relevant libraries and petri nets

```python
import pm4py
import os
```

- load in the intresting petri nets 

```python
heuristic_1_net, heuristic_1_initial_marking, heuristic_1_final_marking = pm4py.read_pnml("../petrinets/heuristic_variants/petri_model_1.pnml")
heuristic_1_bpmn_graph = pm4py.convert_to_bpmn(heuristic_1_net, heuristic_1_initial_marking, heuristic_1_final_marking)
pm4py.view_bpmn(heuristic_1_bpmn_graph)
```
